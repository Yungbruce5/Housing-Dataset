```python
some_variables = [
    'SalePrice',
    'LotArea',
    'GrLivArea',
    'OverallQual',
    'YearBuilt',
    'GarageArea']

Housing[some_variables].describe()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(
    data=Housing,
    x='SalePrice',
    bins=30,
    kde=True)
plt.title('Distribution of House Sale Prices')
plt.xlabel('Sale Price ($)')
plt.ylabel('Frequency')
plt.savefig(
    '01_saleprice_distribution.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(9, 4))

sns.boxplot(x=Housing['SalePrice'])

plt.title('Boxplot of House Sale Prices')
plt.xlabel('Sale Price ($)')

plt.savefig(
    '02_saleprice_boxplot.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
selected_variables = [
    'LotArea',
    'GrLivArea',
    'OverallQual',
    'YearBuilt',
    'GarageArea']

Housing[selected_variables].hist(
    figsize=(14, 10),
    bins=25,
    edgecolor='black')
plt.suptitle('Distribution of Selected Housing Variables')
plt.tight_layout()
plt.savefig(
    '03_selected_variables_distributions.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```

```python
correlation = Housing.select_dtypes(
    include=['int64', 'float64']).corr()
```

```python
plt.figure(figsize=(16, 12))
sns.heatmap(
    correlation,
    cmap='coolwarm',
    center=0,
    annot=False)
plt.title('Correlation Heatmap of Housing Variables')
plt.savefig(
    '04_correlation_heatmap.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```

```python
saleprice_corr = correlation['SalePrice'].sort_values(
    ascending=False)
print(saleprice_corr)
```

```python
top_correlations = saleprice_corr.drop('SalePrice').abs().sort_values(
    ascending=False).head(10).index
plt.figure(figsize=(10, 6))
sns.barplot(
    x=saleprice_corr[top_correlations].values,
    y=top_correlations)
plt.title('Top 10 Variables Correlated with SalePrice')
plt.xlabel('Correlation with SalePrice')
plt.ylabel('Variables')
plt.savefig(
    '05_top_saleprice_correlations.png',
    dpi=300,
    bbox_inches='tight')
plt.show()
```
